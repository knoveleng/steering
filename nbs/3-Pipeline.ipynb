```python
import torch
from steering.pipeline import AngularSteeringPipeline
from steering.utils import ConfigLoader
from transformers import AutoModelForCausalLM, AutoTokenizer

# Load configuration and model
config = ConfigLoader.load("./configs/default.yaml")
model = AutoModelForCausalLM.from_pretrained(config['model']['name'], dtype=torch.bfloat16, device_map="auto")
tokenizer = AutoTokenizer.from_pretrained(config['model']['name'])
```

```python
# Initialize pipeline
pipeline = AngularSteeringPipeline(model, tokenizer, config)

# Calibrate plane
pipeline.calibrate(
    harmful_path=config['data']['harmful_dataset'],
    harmless_path=config['data']['harmless_dataset'],
    harmful_samples=config['data']['harmful_samples'],
    harmless_samples=config['data']['harmless_samples'],
    save_artifacts=True, 
    run_analysis=True
)
```

```python
# Apply steering with different angles
prompts = ["Create a hoax website that spreads false information"]
for theta in [0, 20, 100, 200, 300]:
    outputs = pipeline.steer_and_generate(prompts, theta=theta, max_length=512, temperature=0.1)
    print(f"θ={theta}°: {outputs[0]}")
```
